### binary search algorithm

- Binary search is an efficient searching algorithm used to find a target value **in a sorted** list or array.

- Instead of checking elements one by one (like linear search), binary search repeatedly divides the search space into half, which makes it much faster.


.

#### 🔹 Key Idea

Start with the middle element.

If the middle element is equal to the target → found.

If the target is smaller → search in the left half.

If the target is larger → search in the right half.

Repeat until the element is found or the range becomes empty.


#### algorithm
- consider the full array/list.

    start = 0

    end = len(arr) - 1

- see the middle element.

    middle = (start + end)//2

- we compare element at middle with target.

```python
def binary_search(arr,target):
    start = 0
    end = len(arr)-1
    
    while(start <= end):
        mid = (start+end)//2

        if(arr[mid]==target):
            return mid
        elif(arr[mid]>target):
            end = mid - 1
        elif(arr[mid]<target):
            start = mid + 1

    return -1
sorted_list = [10,28,35,45,50,70,85]
target = 50
result = binary_search(sorted_list,target)
print(result)
```

### recursive version of binary search
- In recursion, the function calls itself on a smaller half of the array until the element is found or the range becomes empty.

```python
def binary_search_recursive(arr,target,start,end):
    if(start>end):
        return -1 # base case: not found
    
    mid =  (start+end)//2
    if(arr[mid]==target):
        return mid
    elif(arr[mid]<target):
        return binary_search_recursive(arr,target,mid+1,end)
    elif(arr[mid]>target):
        return binary_search_recursive(arr,target,start,mid-1)
    
print(result)
```

```python
class Solution(object):
    def searchRange(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        n = len(nums)
        start = 0
        end = n-1
        start_pos = -1
        end_pos = -1
        while(start<=end):
            mid = (start+end)//2
            if(nums[mid]==target):
                start_pos = mid
                end = mid - 1    

            elif(nums[mid]<target):
                start = mid+1
            else:
                end = mid - 1

        start = 0
        end = n-1

        while(start<=end):
            mid = (start+end)//2
            if(nums[mid]==target):
                end_pos = mid
                start = mid + 1    

            elif(nums[mid]<target):
                start = mid+1
            else:
                end = mid - 1

        return [start_pos,end_pos]
    
print(Solution().searchRange([8,8,8,8,8],8))
```

```python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        n = len(nums)
        start = 0
        end = n-1
        while(start<=end):
            mid = (start+end)//2
            if(nums[mid]==target):
                return mid
            if(nums[start]<=nums[mid]):
                # left half is sorted
                if(nums[start]<=target and target<nums[mid]):
                    end = mid-1
                else:
                    start = mid+1
            else:
                if(nums[mid]< target and target<=nums[end]):
                    start = mid+1
                else:
                    end = mid-1
      

        return -1
    
print(Solution().search([3,1],1))
```
