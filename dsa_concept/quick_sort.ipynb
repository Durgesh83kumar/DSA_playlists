## Quick Sort Algorithm

✅ **Quick Sort** is a fast and efficient sorting algorithm based on the **Divide and Conquer** strategy. It works by selecting a **pivot element** and partitioning the array so that:

Elements **less than the pivot** go to the left

Elements **greater than the pivot** go to the right
Then it recursively sorts the left and right subarrays.

### 🔹 How Quick Sort Works (Step-by-Step)

Example:
Array = [5, 2, 9, 1, 7]

1️⃣ Choose a pivot → (let’s pick last element 7)
2️⃣ Partition the array:

Left side (< 7): [5, 2, 1]

Right side (> 7): [9]

So array becomes: [5, 2, 1, 7, 9]

3️⃣ Recursively apply quick sort on [5, 2, 1] and [9]

Final sorted array: [1, 2, 5, 7, 9]

```python
def partitionFunction(arr,s,e):
    pivot = arr[e]
    i = s
    rightposition = s
    while(i<e):
        if(arr[i]<pivot):
            rightposition+=1
        i+=1
    arr[rightposition],arr[e] = arr[e],arr[rightposition]

    pivotIndex = rightposition

    # Now make sure that everything samller than pivot to left and greater than pivot to right
    
    start = s
    end = e
    while(start<pivotIndex and end>pivotIndex):
        if(arr[start]<pivot):
            start += 1
        elif(arr[end]>=pivot):
            end -= 1
        else:
            arr[start],arr[end] = arr[end], arr[start]
            start += 1
            end -= 1

    return pivotIndex

# arr = [3,6,7,2,1,4,5]
# print(partitionFunction(arr,0,len(arr)-1))

def quicksort(arr,s,e):
    if(s>=e):
        return arr

    pivotIndex = partitionFunction(arr,s,e)

    quicksort(arr,s,pivotIndex-1)
    quicksort(arr,pivotIndex+1,e)

    return arr

arr = [3,6,7,2,1,4,5]

print(quicksort(arr,0,len(arr)-1))
```
