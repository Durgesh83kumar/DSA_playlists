```python
# question 1

class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        dictionary = {}
        for i in range(len(nums)):
            current = nums[i]
            needed = target - current
            if(needed in dictionary):
                return [dictionary[needed],i]
            else:
                dictionary[current] = i

s = Solution()
print(s.twoSum([2,7,11,15],9))
print(s.twoSum([3,2,4],6))
print(s.twoSum([3,2],6))
```

```python
class Solution(object):
    def isPalindrome(self, x):
        x = str(x)
        start = 0
        end = len(x)-1
        while(start<end):
            if(x[start]!=x[end]):
                return False
            
            start += 1
            end -= 1

        return True
```

```python
# remove duplicate from sorted linked list
# Definition for singly-linked list.
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution(object):
    def deleteDuplicates(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        temp = head

        while(temp is not None) and (temp.next is not None):
            if temp.val == temp.next.val:
                temp.next = temp.next.next
            else:
                temp = temp.next
        
        return head
```

### 88. Merge Sorted Array

```python
class Solution(object):
    def merge(self, nums1, m, nums2, n):
        """
        :type nums1: List[int]
        :type m: int
        :type nums2: List[int]
        :type n: int
        :rtype: None Do not return anything, modify nums1 in-place instead.
        """
        i = m-1
        j = n-1
        k = m+n-1
        while(i>=0 and j>=0):
            if(nums1[i]>=nums2[j]):
                nums1[k] = nums1[i]
                i = i - 1
                k = k - 1
            else:
                nums1[k] = nums2[j]
                j = j - 1
                k = k - 1

        while(j>=0):
            nums1[k] = nums2[j]
            j = j - 1
            k = k - 1
```

### 169. Majority Element

```python
class Solution(object):
    def majorityElement(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        majority = nums[0]
        count = 1

        for i in range(1,len(nums)):
            if(nums[i]==majority):
                count += 1
            else:
                count -= 1
                if(count==0):
                    majority = nums[i]
                    count = 1
            
        return majority

        
```

```python
class Solution(object):
    def removeDuplicates(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        if(len(nums)<=2):
            return len(nums)
        
        else:        
            k = 2
            for i in range(2,len(nums)):
                if(nums[i]!=nums[k-2]):
                    nums[k] = nums[i]
                    k = k + 1

        return k
```

### 121. Best Time to Buy and Sell Stock

#### Key Idea
- We must buy before we sell.
- We don’t compare all pairs of days.
- We carry best info from the past while moving left → right.
#### What to track
- **min_price** → cheapest price seen so far (best day to buy)
- **max_profit** → best profit seen so far
#### Steps while looping

For each day’s price:

1. Calculate profit if sold today:
        **profit_today = today_price - min_price**
2. Update **max_profit** if this profit is larger.
3. Update **min_price** if today’s price is smaller.
#### Important Order
- First check profit.
- Then update minimum price.
#### Edge Case
- If prices list is empty → return 0.
#### Time Complexity
- O(n) — single loop
- No nested loops, no extra list

```python
class Solution(object):
    def maxProfit(self, prices):
        """
        :type prices: List[int]
        :rtype: int
        """
        if not prices:  # Empty list → treated as False → not makes it True → if runs.
            return 0
    
        min_price = prices[0]
        max_profit = 0
        for i in range(1, len(prices)):
            profit_today = prices[i] - min_price
            if(profit_today>max_profit):
                max_profit = profit_today
            if(prices[i]<min_price):
                min_price = prices[i]
            
      
        return max_profit
```

### 122. Best Time to Buy and Sell Stock II

#### What they are indirectly telling you

They are NOT saying:

-    Find one best buy and one best sell.

They ARE saying:

-    Take profit every time price goes up.

#### Hidden hint inside wording

Because they allow **multiple transactions**, the problem becomes:

-    Capture every increase between consecutive days.

That’s why this thinking works in code:

-    If tomorrow price > today price → take that profit

#### Mental summary in one sentence

-   “Whenever price rises from one day to the next, add that difference to profit.”

```python
class Solution(object):
    def maxProfit(self, prices):
        """
        :type prices: List[int]
        :rtype: int
        """
        if not prices:
            return 0
        profit = 0

        for i in  range(0, len(prices)-1):
            if(prices[i+1]>prices[i]):
                profit = profit + (prices[i+1] - prices[i])
                
        return profit
```

### 55 Jump Game

#### 💡 Core Idea (Greedy)

We do not simulate jumps.

We track a boundary:

-   maxReach = farthest index reachable so far

#### 🧠 Mental Model

Imagine coloring the array from left to right.

- Colored = reachable
- Uncolored = unreachable (game over if you step here)
#### ✅ Algorithm Thinking

For every index i:

1. If i is outside the colored area → **stuck**

-   if i > maxReach → return False

2. Expand the colored area:

-   maxReach = max(maxReach, i + nums[i])

```python
class Solution(object):
    def canJump(self, nums):
        """
        :type nums: List[int]
        :rtype: bool
        """
        maxReach = 0
        for i in range(len(nums)):
            if(i>maxReach):
                return False
            maxReach = max(maxReach, i + nums[i])

        if(maxReach>=len(nums)-1):
            return True

        return False








        
        # maxReach = 0
        
        # for i in range(len(nums)):
        #     if i > maxReach:
        #         return False
            
        #     maxReach = max(maxReach, i + nums[i])
        
        # return True
```

### 45. Jump Game II

**Why this works (quick confirmation)**
- farthest → how far the **next jump** can take you
- currentEnd → boundary of the **current jump**
- When i == currentEnd → you finished scanning this jump’s range → must jump
- Increase jumps, move window to farthest

```python
class Solution(object):
    def jump(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        jumps = 0
        currentEnd = 0
        farthest = 0
        
        for i in range(len(nums)-1):
            farthest = max(farthest, i+nums[i])

            if(i == currentEnd):
                jumps = jumps + 1
                currentEnd = farthest

        return jumps
```
