🔹 What is Bubble Sort?

Bubble sort is a simple sorting algorithm that repeatedly compares adjacent elements and swaps them if they are in the wrong order.

Big elements “bubble up” to the end of the list, just like bubbles rise to the top — that’s where the name comes from.

🔹 How It Works (Idea)

For an array:

[5, 1, 4, 2, 8]

Pass 1:

Compare 5 & 1 → swap → [1, 5, 4, 2, 8]

Compare 5 & 4 → swap → [1, 4, 5, 2, 8]

Compare 5 & 2 → swap → [1, 4, 2, 5, 8]

Compare 5 & 8 → no swap

After pass 1 → largest element 8 is in correct place.

Repeat passes until the array is sorted.

```python
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-1-i):
            if(arr[j]>arr[j+1]):
                arr[j],arr[j+1] = arr[j+1],arr[j]

    return arr

print(bubble_sort([5,1,2,4,9]))
```
