## Linear Search Algorithm (Sequential Search)

Linear search is the simplest searching algorithm. It checks each element in a list one by one until the target value is found or the list ends.

🔹 How It Works

Start from the first element of the list.

Compare it with the target value.

If it matches → return the index (or True).

If not → move to the next element.

Repeat until:

The element is found, or

The entire list is checked (then return -1 or False).

```python
def linear_search(arr, target):
    for index in range(0,len(arr)):
        if (arr[index]==target):
            return index
    
    return -1
    
nums = [1,20,30,23,59]
result = linear_search(nums,23)
print(result)
```

## Recursive Linear Search – Idea

Instead of using a loop, we:

Compare the current index element with the target.

If it matches → return the index.

If not → call the function again for the next index.

If we reach the end of the list → return -1.

```python
def linear_search_recursive(arr,target,index=0):
    # base case: if index reaches end of list
    if (index==len(arr)):
        return -1

    # if element found
    if(arr[index]==target):
        return index
    
    # recursive call for next index
    return linear_search_recursive(arr,target,index+1)

nums = [10,25,30,45,50]
result = linear_search_recursive(nums,30)
# print(result)

if result != -1:
    print("Element found at index: ",result)
else:
    print("Element not found")
```

### 🔹 1. Linear Search for a Character in a String

We check each character one by one until we find the target character.

📌 Example

String: "hello world"

Target: 'o'

Output: 4 (first occurrence)

```python
def linear_search_char(s,target):
    for i in range(len(s)):
        if(s[i]==target):
            return i

    return -1

text = "hello world, I am Durgesh Kumar"
result = linear_search_char(text,'I')
if(result != -1):
    print("character found at index: ",result)
else:
    print("character not found")
```

### 🔹 2. Linear Search for a Word in a String

We split the string into words and then apply linear search.

📌 Example

String: "the quick brown fox"

Target: "brown"

Output: 2

```python
def linear_search_word(sentence,target):
    words = sentence.split() # it covert into list of words
    for i in range(len(words)):
        if (words[i]==target):
            return i
    
    return -1

text = "hello world, I am Durgesh Kumar"
result = linear_search_word(text,"am")
print(result)
```

### 
🔹 3. Find All Occurrences of a Character (Bonus)

Sometimes interviewers ask this 👇



```python
def linear_search_all_chars(s,target):
    indices = []
    for i in range(len(s)):
        if(s[i]==target):
            indices.append(i)

    return indices

text = "hello world, I am Durgesh Kumar"
result = linear_search_all_chars(text,'E')
print(result)
```
